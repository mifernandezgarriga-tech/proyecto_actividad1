## 4. Casos de prueba

Se prueban algunas funciones por separado para comprobar su funcionamiento.

```python
# Buscar una columna existente y una inexistente
print(buscar_columna("EDAD"))
print(buscar_columna("NO_EXISTE"))
```

```python
# Aplicar un umbral de completitud del 90%
seleccion = columnas_de_rol("investigador")
resultado = aplicar_umbral(seleccion, 90)

print("Antes del umbral:")
for columna in seleccion:
    print(columna["nombre"])

print("Con umbral 90:")
for columna in resultado:
    print(columna["nombre"])
```

```python
# Ordenar por nombre ascendente
resultado = ordenar_columnas(COLUMNAS, criterio="nombre", direccion="A")
for columna in resultado:
    print(columna["nombre"])

# Ordenar por completitud descendente
resultado = ordenar_columnas(COLUMNAS, criterio="completitud", direccion="B")
for columna in resultado:
    print(columna["nombre"], columna["completitud"])
```

## 5. Prueba de un rol inexistente

Si se intenta generar un informe con un rol que no existe, el programa informa el error.

```python
try:
    generar_informe("gerente")
except ValueError as error:
    print("Error:", error)
```

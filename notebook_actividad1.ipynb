## 4. Casos de prueba puntuales

Se prueban las funciones auxiliares por separado, para verificar que
cada pieza (búsqueda, filtro por umbral, orden) funciona antes de
combinarlas en `generar_informe`.

```python
# Caso de prueba: buscar una columna existente y una inexistente
print(buscar_columna("EDAD"))
print(buscar_columna("NO_EXISTE"))
```

```python
# Caso de prueba: aplicar un umbral de completitud del 90%
seleccion = columnas_de_rol("investigador")
print("Antes del umbral:", [c["nombre"] for c in seleccion])
print("Con umbral 90:", [c["nombre"] for c in aplicar_umbral(seleccion, 90)])
```

```python
# Caso de prueba: ordenar por nombre ascendente vs. por completitud descendente
print([c["nombre"] for c in ordenar_columnas(COLUMNAS, criterio="nombre", direccion="A")])
print([c["nombre"] for c in ordenar_columnas(COLUMNAS, criterio="completitud", direccion="D")])
```

## 5. Manejo de errores: rol inexistente

`generar_informe` levanta un `ValueError` claro si se pide un rol que
no está en `ROLES`, en vez de fallar de forma confusa.

```python
try:
    generar_informe("gerente")
except ValueError as e:
    print("Error controlado:", e)
```
